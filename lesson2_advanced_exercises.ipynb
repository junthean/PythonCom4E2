## Exercise 1: Numeric analysis

- Create one `int`, one `float` (scientific notation), and one `complex` number.
- Print each value, type, and (for complex) real/imag parts.
- Convert int->float and float->int, then print results.

```python
# 1. Create variables
num_int = 42
num_float = 3.5e4
num_complex = 2 + 5j

# 2. Print value, type, and real/imag parts
print("Integer:", num_int, type(num_int))
print("Float (scientific):", num_float, type(num_float))
print("Complex:", num_complex, type(num_complex), f"real={num_complex.real}, imag={num_complex.imag}")

# 3. Conversions
int_to_float = float(num_int)
float_to_int = int(num_float)

print("int -> float:", int_to_float, type(int_to_float))
print("float -> int:", float_to_int, type(float_to_int))
```

## Exercise 2: Safe casting challenge

- Given `values = ["25", "3.14", "-8", "hello"]`, decide how each text should be converted.
- Use string methods (such as `isdigit()`, `replace()`, `lstrip()`) and `if/elif/else` to check each value.
- Convert to `int` if it is a whole number, to `float` if it is a decimal number, otherwise keep it as a string.
- Print each converted value together with its type.

```python
values = ["25", "3.14", "-8", "hello"]
converted = []

for val in values:
    # Check if whole number (supports negative integers)
    if val.lstrip("-").isdigit():
        result = int(val)
    # Check if decimal number (contains single decimal point)
    elif val.count(".") == 1 and val.lstrip("-").replace(".", "", 1).isdigit():
        result = float(val)
    # Otherwise keep as string
    else:
        result = val
    converted.append(result)
    print(f"Original: {val:<8} | Converted: {str(result):<8} | Type: {type(result)}")
```

## Exercise 3: String slicing matrix

- Use `text = "DataSciencePython"`.
- Print slices: first 4 chars, last 6 chars, every 2nd char, reversed text.
- Extract and print `Science` from the string using slicing only.

```python
text = "DataSciencePython"

# Slices
print("First 4 chars  :", text[:4])
print("Last 6 chars   :", text[-6:])
print("Every 2nd char :", text[::2])
print("Reversed text  :", text[::-1])

# Extract 'Science' using slicing only (indices 4 to 11)
science = text[4:11]
print("Extracted text :", science)
```

## Exercise 4: String transformation pipeline

- Start with `raw = "   python,java,c++ , go  "`.
- Strip the outer spaces and split the text by comma.
- Print the split list.
- Loop through the list, strip each language, uppercase it, and print it.

```python
raw = "   python,java,c++ , go  "

# Strip outer spaces and split by comma
languages = raw.strip().split(",")
print("Split list:", languages)

print("\nProcessed languages:")
# Loop through the list, strip each language, uppercase it, and print it
for lang in languages:
    print(lang.strip().upper())
```

## Exercise 5: Formatting and escaping

- Create variables `product`, `price`, and `discount`.
- Print original and discounted price using f-string with 2 decimals.
- Print one line containing escaped quotes and another with newline+tab formatting.

```python
# Create variables
product = "Wireless Headphones"
price = 129.99
discount = 0.20  # 20% discount

discounted_price = price * (1 - discount)

# Original and discounted price using f-string with 2 decimals
print(f"Product: {product}")
print(f"Original Price: ${price:.2f}")
print(f"Discounted Price: ${discounted_price:.2f}")

# Escaped quotes
print("She said, \"Python formatting is powerful!\"")

# Newline and tab formatting
print("\nReceipt Summary:\n\tItem:\tWireless Headphones\n\tTotal:\t$" + f"{discounted_price:.2f}")
```
