## Exercise 5: Type-driven expression building

- Create variables for name (str), age (int), height (float), active (bool).
- Print all four variables in one line using commas inside `print()`.
- Create `first_name` and `last_name`, then print the full name using `+` and a space.
- Add `age` and `height`, print the result and its type.
- Print the type of every variable at the end.

```python
# Write your solution here
name, age, height, active = "Thean", 20, 1.75, True

print("Profile:", name, age, height, active)

first_name, last_name = "Lun", "Sithean"
print("Full Name:", first_name + " " + last_name)

sum_val = age + height
print(f"Sum: {sum_val} ({type(sum_val).__name__})")

print("\n--- Types ---")
for var_name, var in [("name", name), ("age", age), ("height", height), ("active", active)]:
    print(f"{var_name:<8} : {type(var).__name__}")

```
