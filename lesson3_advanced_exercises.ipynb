## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
# Create booleans
p = True
q = False

# Logical operations
print(f"p = {p}, q = {q}")
print("p and q :", p and q)
print("p or q  :", p or q)
print("not p   :", not p)
print("not q   :", not q)

# Compound expression with parentheses
compound = (p or q) and not (p and q)
print("(p or q) and not (p and q):", compound)
```

```python
a = 12
b = 12
c = 7

print(f"a = {a}, b = {b}, c = {c}\n")

# Comparisons
print("1. a == b      :", a == b)          # Equality
print("2. a != c      :", a != c)          # Inequality
print("3. a > c       :", a > c)           # Greater than
print("4. c < a       :", c < a)           # Less than
print("5. a >= b      :", a >= b)          # Greater than or equal to
print("6. c <= b      :", c <= b)          # Less than or equal to
print("7. c < a <= b  :", c < a <= b)      # Chained comparison
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Operator precedence in Python: not > and > or
x = True
y = False
z = False

# Without extra parentheses (default precedence: not, then and, then or)
expr_default = x and not y or z

# With explicit parentheses
expr_explicit = (x and (not y)) or z

print("Without extra parentheses :", expr_default)
print("With explicit parentheses :", expr_explicit)
print("Do they match?            :", expr_default == expr_explicit)

# Comment: They match because Python naturally evaluates 'not' before 'and', and 'and' before 'or'.
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# 1. Create two equal lists
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print("Comparing two separate equal lists:")
print("list1 == list2 :", list1 == list2)  # True: values are equal
print("list1 is list2 :", list1 is list2)  # False: different objects in memory

# 2. Assign list1 to another variable
list3 = list1

print("\nComparing list1 with referenced list3:")
print("list1 == list3 :", list1 == list3)  # True: values are equal
print("list1 is list3 :", list1 is list3)  # True: same object in memory

# Short explanation:
# '==' evaluates whether values/contents are equal.
# 'is' evaluates whether two variables reference the exact same object in memory.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
items = ["python", "java", "go", "rust"]
queries = ["python", "c++", "rust", "html"]

for query in queries:
    if query in items:
        print(f"{query}: found")
    elif query not in items:
        print(f"{query}: not found")
```
